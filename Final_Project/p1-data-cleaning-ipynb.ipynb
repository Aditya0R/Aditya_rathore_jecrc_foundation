```python
# Basic Information
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Shape of dataset
print("Number of Rows :", df.shape[0])
print("Number of Columns :", df.shape[1])

# First 5 rows
df.head()
```

```python
missing = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': round(df.isnull().mean()*100,2)
})

missing = missing.sort_values(by='Percentage', ascending=False)

missing
```

We calculate the percentage of missing values because the decision to drop or impute a feature depends on the proportion of missing data, not just the count.

```python
df.drop(['Prospect ID','Lead Number'], axis=1, inplace=True)
```

```python
df.shape
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
for col in df.columns:
    print("="*60)
    print(col)
    print(df[col].value_counts().head())
```

```python
constant_cols = []

for col in df.columns:
    if df[col].nunique() == 1:
        constant_cols.append(col)

print(constant_cols)
```

```python
df.drop(columns=constant_cols, inplace=True)
```

```python
round(df.isnull().mean()*100,2).sort_values(ascending=False)
```

```python
drop_cols = [
'Lead Quality',
'Asymmetrique Activity Index',
'Asymmetrique Profile Index',
'Asymmetrique Activity Score',
'Asymmetrique Profile Score'
]

df.drop(columns=drop_cols, inplace=True)
```
